# Archive.org → Prophet Drive importer
Mounts Google Drive, downloads public Archive.org EPUB files when available, falls back to PDF, validates downloads, skips existing valid copies, and writes `status/archive_import_report.json`.


```python
from google.colab import drive
drive.mount('/content/drive')
import requests
u='https://raw.githubusercontent.com/houseofwordslangue-dev/Prophet/main/scripts/import_archive_to_drive.py'
r=requests.get(u, timeout=60); r.raise_for_status()
open('/content/import_archive_to_drive.py','wb').write(r.content)
%run /content/import_archive_to_drive.py
```
