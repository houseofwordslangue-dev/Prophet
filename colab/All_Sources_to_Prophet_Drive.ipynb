# Prophet Library — all-source importer
Imports catalogue-matched books into `Prophet-Library-Ingestion/books` from Internet Archive, Masaha, Wikisource (AR/EN/FR), Project Gutenberg (EN/FR), and configured OpenITI exact sources. EPUB is preferred; valid existing EPUBs are skipped. Reports are written to `Prophet-Library-Ingestion/status/`.


```python
!pip -q install beautifulsoup4
from google.colab import drive
drive.mount('/content/drive')
```

```python
import requests, pathlib
BASE='https://raw.githubusercontent.com/houseofwordslangue-dev/Prophet/main/scripts/'
for name in ['import_archive_to_drive.py','import_multisource_to_drive.py']:
    r=requests.get(BASE+name, timeout=60); r.raise_for_status()
    pathlib.Path('/content/'+name).write_bytes(r.content)
print('Importers downloaded.')
```

```python
# 1) Internet Archive
%run /content/import_archive_to_drive.py
```

```python
# 2) Masaha + Wikisource + Project Gutenberg + configured OpenITI
%run /content/import_multisource_to_drive.py --max-auto 150
```

## Reports
- `Prophet-Library-Ingestion/status/archive_import_report.json`
- `Prophet-Library-Ingestion/status/multisource_import_report.json`

A report status of `SKIPPED_VALID_EPUB_EXISTS` means the canonical Drive folder already contains a usable EPUB. `NO_MATCH`, `NO_IMPORT`, or `ERROR` items are retained for later source-specific repair; they are not silently accepted.
